```python
import requests
import csv

def save_api_csv_response(url, filename):
    """Fetches data from the given API URL and saves it as a CSV file.

    Args:
        url (str): The URL of the API endpoint.
        filename (str): The desired filename for the CSV file.
    """

    response = requests.get(url)

    if response.status_code == 200:
        data = response.json()

        # Extract field names from the first data item
        fieldnames = data[0].keys()

        with open(filename, 'w', newline='') as csvfile:
            writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(data)


        print(f"Data saved to {filename}")
    else:
        print(f"Error fetching data: {response.status_code}")

# Example usage:
url = "https://jsonplaceholder.typicode.com/posts"  # Replace with your desired API URL
filename = "posts.csv"

save_api_csv_response(url, filename)
```

```python
import json
import snowflake.Connector


def connect_to_snowflake(config_file):
  """Connects to Snowflake using a JSON configuration file.

  Args:
    config_file (str): Path to the JSON configuration file.

  Returns:
    snowflake.connector.connection: A Snowflake connection object.
  """

  with open(config_file, 'r') as f:
    config = json.load(f)

  conn = snowflake.connector.connect(
      user=config['username'],
      password=config['password'],
      account=config['account_identifier'],
      warehouse=config['warehouse'],
      database=config['database'],
      region=config['region']
  )

  return conn

# Example usage:
config_file = "C:\\Users\\gaura\\OneDrive\\Documents\\DataFlowPro\\configs\\snowflakeConnConfig.json"
conn = connect_to_snowflake(config_file)

# Use the connection to execute queries
with conn.cursor() as cur:
  cur.execute("SELECT CURRENT_USER()")
  for row in cur:
    print(row)

conn.close()
```
